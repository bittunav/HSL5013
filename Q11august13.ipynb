```python
import pandas as pd
import matplotlib.pyplot as plt
import os
print(os.getcwd())
df=pd.read_csv("datadino.csv")

for d,g in df.groupby("dataset"):
    plt.figure()
    plt.scatter(g["x"],g["y"])
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(f"Dataset {d}")
    plt.show()

print(df.groupby("dataset")[["x","y"]].mean())
print(df.groupby("dataset")[["x","y"]].std())
print(df.groupby("dataset")[["x","y"]].corr())
```
