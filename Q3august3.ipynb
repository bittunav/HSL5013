```python
import numpy as np
import matplotlib.pyplot as plt

demand = np.array([0, 50, 100, 200])
prob = np.array([0.05, 0.40, 0.30, 0.25])

x_values = np.arange(0, 301)

profits = []
for x in x_values:
    expected_profit = 0.0
    for d, p in zip(demand, prob):
        profit = 100*min(x, d) - 20*x - 5*max(x-d, 0)
        expected_profit += p*profit
    profits.append(expected_profit)

plt.figure(figsize=(9,6))
plt.plot(x_values, profits)
plt.xlabel("Number of kits produced")
plt.ylabel("Expected weekly profit")
plt.title("Expected Weekly Profit")
plt.grid(True)
plt.savefig("expected_profit.png", dpi=300, bbox_inches="tight")
plt.show()

best_x = x_values[np.argmax(profits)]
best_profit = max(profits)

print("Optimal production =", best_x)
print("Maximum expected profit =", best_profit)
```
